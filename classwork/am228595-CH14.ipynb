```python
import numpy as np
import matplotlib.pyplot as plt

# Function and true derivative
f = lambda x: np.sin(x)
df_true = lambda x: np.cos(x)

# Interval and number of points
a = 0
b = 2*np.pi
N = 200

# Step size
h = (b - a) / (N - 1)

# Grid points
x = np.linspace(a, b, N)

# Central Difference
df_central = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference
df_forward = (f(x + h) - f(x)) / h

# Backward Difference
df_backward = (f(x) - f(x - h)) / h

# -----------------------------
# Plotting
# -----------------------------
plt.figure(figsize=(12, 10))

# True derivative
plt.subplot(2, 2, 1)
plt.plot(x, df_true(x), label="cos(x)", color="black")
plt.title("True Derivative: cos(x)")
plt.grid(True)

# Forward Difference
plt.subplot(2, 2, 2)
plt.plot(x, df_forward, label="Forward Difference", color="blue")
plt.title("Forward Difference Approximation")
plt.grid(True)

# Backward Difference
plt.subplot(2, 2, 3)
plt.plot(x, df_backward, label="Backward Difference", color="red")
plt.title("Backward Difference Approximation")
plt.grid(True)

# Central Difference
plt.subplot(2, 2, 4)
plt.plot(x, df_central, label="Central Difference", color="green")
plt.title("Central Difference Approximation")
plt.grid(True)

plt.tight_layout()
plt.show()
```
